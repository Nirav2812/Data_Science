```python
class Tops:
    def __init__(self , name , email_id , phone_nu , age):
        self.name = name
        self.email_id = email_id
        self.phone_num = phone_nu
        self.age = age

    def return_detailes(self):
        return self.name, self.email_id , self.phone_num , self.age
```
