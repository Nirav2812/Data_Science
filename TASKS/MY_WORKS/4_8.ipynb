```python
from abc import ABC, abstractclassmethod
class Vehical(ABC):
    @abstractclassmethod
    def start(self):
        pass    

class Car(Vehical):
    def start(Self):
        print("car started")

class bike(Vehical):
    def move(Self):
        print("bike is moving")
class bus(Vehical):
    pass
```

```python
bu = bus()
bu.start()
```
