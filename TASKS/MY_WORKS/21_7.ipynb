```python
# lambda - anonymous function
```

```python
def tast(a,b):
    return(a*b)
```

```python
tast(5,8)
```

```python
a = lambda a,b,c : a*b*c
a(2,6,4)
```

```python
b = lambda c,d : (c*c,d*d)
b(4,8)
```

```python
c = lambda e :(e*9/5) + 32
c(10)
```

```python
# a > b

z = lambda a,b : a if a>b else b
z(4,6)
```

```python
from functools import reduce
u = [1,5,8,4,5,5,7]
reduce(lambda x,y, : x+y ,u)
```

```python
l = [78,46,12,78,56,12,78,56,89]
reduce(lambda x,y : x if x>y else y , l)
```

```python
# filter
l = [4,2,5,7,9,1,3,5]
list(filter(lambda x:x%2==1 , l ))
```

```python
l9 = [8,-9,4,7,-6,-7,-3,7,2]
list(filter(lambda x:x>0,l9 ))
```
