```python
# Lambda Function

# Q1. 
add = lambda a, b: a + b
print(add(10, 20))

# Q2. 
square = lambda n: n ** 2
print(square(5))

# Q3. 
even = lambda n: n % 2 == 0
print(even(10))

# Q4. 
larger = lambda a, b: a if a > b else b
print(larger(10, 20))

# Q5. 
area = lambda l, w: l * w
print(area(10, 5))
```

```python
# map()

# Q6. 
numbers = [1, 2, 3, 4, 5]
print(list(map(lambda x: x * 5, numbers)))

# Q7. 
names = ["nirav", "rahul", "jay"]
print(list(map(str.upper, names)))

# Q8. 
marks = [30, 40, 50, 60]
print(list(map(lambda x: x + 10, marks)))

# Q9. 
products = [" laptop ", " mobile ", " keyboard "]
print(list(map(str.strip, products)))

# Q10. 
celsius = [0, 10, 20, 30]
print(list(map(lambda c: (c * 9/5) + 32, celsius)))
```

```python
# filter()

# Q11. 
numbers = [1, 2, 3, 4, 5, 6]
print(list(filter(lambda x: x % 2 == 0, numbers)))

# Q12. 
marks = [35, 40, 55, 30, 70]
print(list(filter(lambda x: x >= 40, marks)))

# Q13. 
prices = [500, 1500, 2500, 800]
print(list(filter(lambda x: x > 1000, prices)))

# Q14. 
names = ["Nirav", "Rahul", "Amit", "Karan", "Pratik"]
print(list(filter(lambda x: len(x) > 5, names)))

# Q15.
numbers = [6, 8, 12, 15, 18, 20]
print(list(filter(lambda x: x % 2 == 0 and x % 3 == 0, numbers)))
```

```python
# reduce()

from functools import reduce

# Q16. 
numbers = [1, 2, 3, 4, 5]
print(reduce(lambda a, b: a + b, numbers))

# Q17. 
numbers = [1, 2, 3, 4, 5]
print(reduce(lambda a, b: a * b, numbers))

# Q18. 
numbers = [10, 25, 5, 40, 15]
print(reduce(lambda a, b: a if a > b else b, numbers))

# Q19. 
numbers = [10, 25, 5, 40, 15]
print(reduce(lambda a, b: a if a < b else b, numbers))

# Q20. 
sales = [10000, 15000, 12000, 18000]
print(reduce(lambda a, b: a + b, sales))


# Q21. 
names = [" nirav ", " rahul ", " jay "]
print(list(map(lambda x: x.strip().upper(), names)))

# Q22. 
salaries = [30000, 55000, 70000, 45000]
print(list(filter(lambda x: x > 50000, salaries)))

# Q23. 
salaries = [30000, 55000, 70000, 45000]
print(reduce(lambda a, b: a + b, salaries))

# Q24.
numbers = [1, 2, 3, 4, 5, 6]
print(list(map(lambda x: x * 2 if x % 2 == 0 else x, numbers)))

# Q25. 
marks = [40, 55, 60, 45, 75]
print(list(map(lambda x: x + 5, filter(lambda x: x > 50, marks))))


# Q26. 
products = [" laptop ", " mobile ", " keyboard ", " mouse "]
print(list(map(lambda x: x.strip().title(), products)))

# Q27. 
prices = [500, 1500, 2500, 800, 3200]
filtered = list(filter(lambda x: x > 1000, prices))
print(filtered)
print(reduce(lambda a, b: a + b, filtered))
```

```python
# Iterator

# Q1. 
numbers = [1, 2, 3, 4, 5]
iterator = iter(numbers)

for x in iterator:
    print(x)
```

```python
# Q2. 
numbers = [1, 2, 3, 4, 5]
iterator = iter(numbers)

while True:
    try:
        print(next(iterator))
    except StopIteration:
        break
```

```python
# Q3. 
numbers = [1, 2, 3, 4, 5]
iterator = iter(numbers)
total = 0

while True:
    try:
        total += next(iterator)
    except StopIteration:
        break

print(total)
```

```python

# enumerate()

# Q1. 
names = ["Nirav", "Rahul", "Jay"]

for index, name in enumerate(names):
    print(index, name)
```

```python
# Q2. 
names = ["Nirav", "Rahul", "Jay"]

for index, name in enumerate(names, start=1):
    print(index, name)
```

```python
# Q3. 
languages = ["Java", "C++", "Python", "JavaScript"]

for index, language in enumerate(languages):
    if language == "Python":
        print(index)
```

```python

# zip()

# Q1. 
names = ["Nirav", "Rahul", "Jay"]
marks = [80, 75, 90]

print(list(zip(names, marks)))

# Q2. 
names = ["Nirav", "Rahul", "Jay"]
marks = [80, 75, 90]

student = dict(zip(names, marks))
print(student)

# Q3. 
names = ["Nirav", "Rahul", "Jay"]
marks = [80, 75, 90]

for name, mark in zip(names, marks):
    print(name, mark)
```

```python
# Generator

# Q1. 
def numbers():
    yield 1
    yield 2
    yield 3

for x in numbers():
    print(x)
```

```python
# Q2. 
def squares(n):
    for i in range(1, n + 1):
        yield i ** 2

for x in squares(5):
    print(x)
```

```python



# Q3.
def even_numbers(n):
    for i in range(1, n + 1):
        if i % 2 == 0:
            yield i

for x in even_numbers(10):
    print(x)
```
