# single inheritance

```python
class Parents:
    def test(self):
        print("this is parents class")

class Child(Parents):
    def test1(self):
        print("this is parents class")
    
```

```python
cd = Child()
cd.test1()
```

```python
cd.test()
```

```python
par = Parents()
par.test()
```

# multiple inheritance

```python
class dad():
    def working(self):
        return "learn to work"
    
class mom():
    def washing(self):
        return "learn washing your cloths"

class child(dad,mom):
    def masti(self):
        return "learn how to masti"
```

```python
da = dad()
mo = mom()
chi = child()
```

```python
chi.washing()
```

```python
chi.working()
```

```python
chi.masti()
```

```python
# class = animaal
# method = sound
# create child class dog and cat that override the sound method

class animal():
    def sound(self):
        print("animal sound")

class dog(animal):
    def sound(self):
            print("dog sound")
class cat(animal):
    def sound(self):
            print("cat sound")
```

```python
dg = dog()
dg.sound()
```

```python
ct = cat()
ct.sound()
```
