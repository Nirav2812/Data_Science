```python
# 0,1,1,2,3,5,8,12,...
def fib_ser(n):
    a,b = 0,1
    for i in range(n):
        yield a
        a,b = b,a+b

# a = 0 b =1
# a = 1 b =1
# a = 1 b =2
# a = 2 b =3
# a = 3 b =5
# a = 5 b =8 
```

```python
a = fib_ser(15)
for i in a:
    print(i)
```
