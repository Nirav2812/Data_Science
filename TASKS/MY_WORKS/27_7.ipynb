```python
try :
    file = open("xyz.txt" , "r")
except Exception as e:   # e = eliyas we dont have to call exception everytime we can call it with the name of e e=petname
    print("this is my except block" , e)

else:
    file.close()
    print(" xyzzz")
finally:
    print("finally")
```

```python
try:
    a = int(input("enter a number a : "))
    b = int(input("enter a number b : "))
    print(a/b)
except ValueError:
    print("this is error")
except ZeroDivisionError:
    print("you can not devide by 0")
else:
    print("successfully print without error")
finally:
    print("program ended")
```
