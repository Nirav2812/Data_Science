```python
import re 
text = "hellouu mittar , your maths score is 69 and your physics score is 67"
```

```python
# syntax : re.findall(pattern,text)
# \d etle digit ne recognise karvu(0-9)
# \d+ one or more digit recignise karse
number = re.findall(r"\d+" , text)
print(number)
```

```python
number = re.findall(r"\w+" , text)
print(number)
```

```python
# re.search - search the pattern , simmilet words
number = re.search(r"\d+" , text)
number.group()
```

```python
text1 = "helow , python my fav number is 69"
number = re.search(r"\w+" , text1)
number.group()
```

```python
# import regex symbols
"""
\d,\w
\D-non digit
\W-non words
\s - space
\S - non space
+ = one or more
* = zero or more
? = zero or one
{a,n} - between a and n
^ - start with ^hello
$ - end with hello$
"""

import re

date = "25-july-2005"
re.findall(r"\d[0-9]{2}-\d[a-zA-Z0-9]{9}-\d[0-9]{4}" , date)
```
