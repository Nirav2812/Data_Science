```python
# 1. Mean

marks = [10, 20, 30, 40, 50]

mean = sum(marks) / len(marks)

print("Mean =", mean)
```

```python
# 2. Median

import statistics

data = [10, 20, 30, 40, 50]

median = statistics.median(data)

print("Median =", median)
```

```python
# 3. Mode

import statistics

data = [10, 20, 20, 30, 40]

mode = statistics.mode(data)

print("Mode =", mode)
```

```python
# 4. Range

data = [10, 20, 30, 40, 50]

range_value = max(data) - min(data)

print("Range =", range_value)
```

```python
# 5. Variance

import statistics

data = [10, 20, 30]

variance = statistics.pvariance(data)

print("Variance =", variance)
```

```python
# 6. Standard Deviation

import statistics

data = [10, 20, 30]

standard_deviation = statistics.pstdev(data)

print("Standard Deviation =", standard_deviation)
```

```python
# 9. Quartile and IQR

import numpy as np

data = [10, 12, 15, 18, 20, 25, 30, 35]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
# 10. Probability

favourable = 1
total = 6

probability = favourable / total

print("Probability =", probability)
```

```python
# 11. Complement Rule

P_A = 0.7

P_not_A = 1 - P_A

print("P(A') =", P_not_A)
```

```python
# 12. Addition Rule

P_A = 0.4
P_B = 0.3
P_A_and_B = 0.1

P_A_or_B = P_A + P_B - P_A_and_B

print("P(A or B) =", P_A_or_B)
```

```python
# 13. Multiplication Rule

P_A = 0.5
P_B = 0.2

P_A_and_B = P_A * P_B

print("P(A and B) =", P_A_and_B)
```

```python
# 14. Bayes' Theorem

P_B_given_A = 0.8
P_A = 0.3
P_B = 0.5

P_A_given_B = (P_B_given_A * P_A) / P_B

print("P(A|B) =", P_A_given_B)
```

```python
# 15. Uniform Distribution

a = 0
b = 10

density = 1 / (b - a)
mean = (a + b) / 2

print("Density =", density)
print("Mean =", mean)
```

```python
# 16. Z-Score

x = 70
mean = 50
sd = 10

z = (x - mean) / sd

print("Z-Score =", z)
```

```python
# 17. Central Limit Theorem

import numpy as np

population = np.arange(1, 101)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

print("Average of sample means =", np.mean(sample_means))
```
