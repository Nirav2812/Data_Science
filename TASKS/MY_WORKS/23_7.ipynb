```python
# factorial 
# 5! = 5*4*3*2*1
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)

fact(5)
```

```python
5*fact(4)
5*4*fact(3)
5*4*3*fact(2)
```

```python
# 1 to n
def sum (n):
    if n ==1:
        return 1
    else:
        return n+sum(n-1)
```

```python
sum(100)
```

```python
# call stack
def greet():
    print("hello")

def demo():
    greet()
    print("welcome")
```

```python
# global keyword - use globall if you want to modify a global variable inside a function
count = 0
def demo1():
    global count
    count+=2

demo1()
demo1()
```
