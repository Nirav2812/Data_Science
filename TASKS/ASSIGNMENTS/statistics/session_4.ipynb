```python
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.plot(range(1, 11), orders, marker='o')
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.title("Daily Orders of Swiggy Delivery Partner")
plt.show()
```

```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]


mean = sum(followers) / len(followers)

followers.sort()
middle = len(followers) // 2
median = followers[middle]

print("Mean =", mean)
print("Median =", median)

if mean > median:
    print("Data is right-skewed")
elif mean < median:
    print("Data is left-skewed")
else:
    print("Data is symmetrical")
```

```python
views = [500, 700, 800, 1000, 1200, 1500, 1800, 2500, 5000, 50000]

mean = sum(views) / len(views)

views.sort()
middle = len(views) // 2
median = (views[middle - 1] + views[middle]) / 2

print("Mean =", mean)
print("Median =", median)
```
