```python
import matplotlib.pyplot as plt


instagram_hours = [1, 2, 2, 3, 4, 4, 5]


posts = [1, 2, 3, 3, 4, 5, 6]

plt.scatter(instagram_hours, posts)

plt.xlabel("Hours spent on Instagram")
plt.ylabel("Number of Posts")
plt.title("Instagram Hours vs Number of Posts")

plt.show()
```

```python
import numpy as np


zomato_orders = [5, 8, 10, 12, 15, 18]

swiggy_orders = [4, 7, 9, 13, 14, 17]


correlation = np.corrcoef(zomato_orders, swiggy_orders)

print("Correlation Matrix:")
print(correlation)

print("Correlation Coefficient:", correlation[0][1])
```

```python
import matplotlib.pyplot as plt

spotify_hours = [1, 2, 2, 3, 4, 5, 6, 7, 8, 9]

marks = [95, 90, 88, 82, 78, 72, 68, 62, 55, 50]

plt.scatter(spotify_hours, marks)

plt.xlabel("Time Spent on Spotify (Hours)")
plt.ylabel("Test Marks")
plt.title("Spotify Time vs Test Marks")

plt.show()
```
