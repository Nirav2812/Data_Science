task 4

Three useful statistical terms for a playlist recommendation feature:

1. Mean (Average)
   Example: If a user listens to 30 songs over 5 days, the mean is 6 songs per day.

2. Median
   Example: If a user's daily listening times are 10, 20, 25, 30, and 60 minutes, the median listening time is 25 minutes.

3. Mode
   Example: If a user listens to Pop, Rock, Pop, Jazz, and Pop, the mode is Pop because it occurs most frequently.
