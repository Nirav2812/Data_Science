```python
# Task 1: Calculate Variance

steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

# Step 1: Calculate mean
mean = sum(steps) / len(steps)
print("Mean =", mean)

# Step 2: Calculate squared deviations
squared_deviations = []

for x in steps:
    deviation = x - mean
    squared = deviation ** 2
    squared_deviations.append(squared)
    print(x, "-", mean, "=", deviation, "->", squared)

# Step 3: Calculate variance
variance = sum(squared_deviations) / len(steps)

print("\nSum of squared deviations =", sum(squared_deviations))
print("Variance =", variance)
```

```python
# Task 2: Calculate Standard Deviation

import math

def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)

    squared_deviations = [(x - mean) ** 2 for x in scores]

    variance = sum(squared_deviations) / len(scores)

    standard_deviation = math.sqrt(variance)

    return round(standard_deviation, 2)


# Example
scores = [80, 85, 90, 75, 95]

result = calculate_standard_deviation(scores)

print("Scores:", scores)
print("Standard Deviation:", result)
```

```python
# Task 3: Compare spending consistency

import math

def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)

    squared_deviations = [(x - mean) ** 2 for x in scores]

    variance = sum(squared_deviations) / len(scores)

    return round(math.sqrt(variance), 2)


friend_A = [200, 200, 200, 200, 200]
friend_B = [100, 300, 150, 400, 50]

sd_A = calculate_standard_deviation(friend_A)
sd_B = calculate_standard_deviation(friend_B)

print("Friend A Standard Deviation:", sd_A)
print("Friend B Standard Deviation:", sd_B)

if sd_A < sd_B:
    print("\nFriend A has more consistent spending.")
    print("Friend A spends the same amount every day, so the variation is zero.")
else:
    print("\nFriend B has more consistent spending.")
```

```python
# Task 4: Salary Variance and Standard Deviation

import math

salary = [28000, 29000, 27000, 60000, 26500, 27500]

# Step 1: Calculate mean
mean = sum(salary) / len(salary)

# Step 2: Calculate squared deviations
squared_deviations = [(x - mean) ** 2 for x in salary]

# Step 3: Calculate variance
variance = sum(squared_deviations) / len(salary)

# Step 4: Calculate standard deviation
standard_deviation = math.sqrt(variance)

print("Salary Data:", salary)
print("Mean Salary:", round(mean, 2))
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(standard_deviation, 2))
```
