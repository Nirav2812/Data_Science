```python
plays = [120, 135, 150, 200, 120, 90, 200]

total = sum(plays)
count = len(plays)

mean = total / count

print("Mean plays:", mean)
```

```python
delivery_times = [30, 25, 40, 35, 30, 45, 30]

delivery_times.sort()

n = len(delivery_times)

median = delivery_times[n // 2]

print("Median delivery time:", median)
```

```python
def most_common_rating(ratings):
    counts = {}

    for rating in ratings:
        if rating in counts:
            counts[rating] += 1
        else:
            counts[rating] = 1

    mode = max(counts, key=counts.get)

    return mode


ratings = [5, 4, 4, 3, 5, 4, 2, 4]

print("Most common rating:", most_common_rating(ratings))
```

```python
from statistics import mean, median, mode

channel_a = [100, 120, 110, 105, 5000]

channel_b = [100, 120, 110, 105, 130]

channel_c = [100, 100, 110, 110, 120]

print("Channel A")
print("Mean:", mean(channel_a))
print("Median:", median(channel_a))
print("Mode:", mode(channel_a))

print("\nChannel B")
print("Mean:", mean(channel_b))
print("Median:", median(channel_b))
print("Mode:", mode(channel_b))

print("\nChannel C")
print("Mean:", mean(channel_c))
print("Median:", median(channel_c))
print("Mode:", mode(channel_c))
```
