```python
# Task 1 — Instagram vs YouTube

# Scenario: We believe that people spend more time on Instagram than on YouTube daily.

# Hypotheses:
# Null Hypothesis (H₀): People do not spend more time on Instagram than on YouTube.
# Alternative Hypothesis (H₁): People spend more time on Instagram than on YouTube.

# In simple form:

# H₀: Instagram time ≤ YouTube time
# H₁: Instagram time > YouTube time
```

```python
# Task 2 — Zomato A/B Test

# Given:

# Group A = ₹350
# Group B = ₹320
# Standard deviation = ₹50
# Number of users in each group = 30
# Significance level = 0.05
```

```python
from scipy.stats import ttest_ind_from_stats

mean_A = 350
std_A = 50
n_A = 30

mean_B = 320
std_B = 50
n_B = 30

t_stat, p_value = ttest_ind_from_stats(
    mean1=mean_A,
    std1=std_A,
    nobs1=n_A,
    mean2=mean_B,
    std2=std_B,
    nobs2=n_B
)

print("t-statistic =", t_stat)
print("p-value =", p_value)


if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Accept the null hypothesis")
```

```python
# Task 3 — A/B Test for WhatsApp Dark Mode

# Suppose WhatsApp introduces a new dark-mode design.

# Control Group

# Users see the old dark-mode design.

# Variant Group

# Users see the new dark-mode design.

# Metric

# We can measure average daily usage time or the percentage of users who continue using the app after enabling dark mode.

# Hypotheses

# H₀: The new dark mode does not increase user engagement.

# H₁: The new dark mode increases user engagement.

# We randomly divide users into two groups and compare their results. If the p-value is below 0.05, we reject H₀ and conclude that the new design has a statistically significant effect.
```
