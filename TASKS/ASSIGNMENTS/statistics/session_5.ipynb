```python
# Task 1 — Population and Sample
# Example 1 — Instagram Followers
# Population: All followers of an Instagram account.
# Sample: 100 randomly selected followers from those followers.
# Example 2 — Zomato Restaurants
# Population: All restaurants listed on Zomato.
# Sample: 50 randomly selected restaurants from the complete list.
# Example 3 — Flipkart Products
# Population: All products available on Flipkart.
# Sample: 100 randomly selected products from the complete product list.
```

```python
import random

# Create 1000 fake user IDs
users = []

for i in range(1, 1001):
    users.append("user" + str(i))

# Select 50 random users
sample = random.sample(users, 50)

# Print selected users
print("Selected Users:")

for user in sample:
    print(user)
```

```python
# Task 3 — Why Use Sampling?

# We use sampling because collecting data from every Swiggy order would take a lot of time, money, 
# and resources. A properly selected sample of orders can give us useful information about delivery 
# times without analyzing every order.
```

```python
# Task 4 — Sampling Bias

# The survey is sent only to users who recently streamed Bollywood songs.

# This introduces selection bias because these users may have different 
# music preferences from the overall Spotify user population. The sample is not representative 
# of all Spotify users because people who do not listen to Bollywood songs are excluded.
```
