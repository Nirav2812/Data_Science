# task 1

```python
class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def show_playlist(self):
        print(self._songs)


playlist = Playlist()

playlist.add_song("Blinding Lights")
playlist.add_song("Starboy")
playlist.add_song("Perfect")

playlist.show_playlist()
```

# task 2

```python
class Product:
    def __init__(self, price):
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, price):
        self._price = price


product = Product(15000)

print("Original Price:", product.get_price())

product.set_price(12000)

print("Updated Price:", product.get_price())
```

# task 3

```python
class Movie:
    def __init__(self, rating):
        self._rating = 0
        self.set_rating(rating)

    def get_rating(self):
        return self._rating

    def set_rating(self, rating):
        if 0 <= rating <= 10:
            self._rating = rating
        else:
            print("Error: Rating must be between 0 and 10.")


movie = Movie(8.5)

print("Rating:", movie.get_rating())

movie.set_rating(9)

print("Updated Rating:", movie.get_rating())

movie.set_rating(15)
```

# task 4

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class Paytm(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Paytm")


class PhonePe(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ₹{amount} using PhonePe")


paytm = Paytm()
phonepe = PhonePe()

paytm.pay(500)
phonepe.pay(750)
```
