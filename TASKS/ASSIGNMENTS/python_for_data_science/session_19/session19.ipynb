# task 4

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price

    def add_item(self, item_name, item_price):
        self.items.append(item_name)
        self.total_price += item_price


order1 = FoodOrder(
    "Domino's",
    ["Margherita Pizza", "Garlic Bread"],
    450
)

order1.add_item("Coke", 60)
order1.add_item("Brownie", 100)

print("Restaurant:", order1.restaurant_name)
print("Items:", order1.items)
print("Updated Total:", order1.total_price)
```

# task 5

```python
class Song:
    def __init__(self, title, artist, duration=0):
        self.title = title
        self.artist = artist
        self.duration = duration

    def play_preview(self):
        print(f"Playing 30-second preview of {self.title} by {self.artist}")


# Duration provided
song1 = Song("Blinding Lights", "The Weeknd", 200)

# Duration not provided
song2 = Song("Starboy", "The Weeknd")

print(song1.title)
print(song1.artist)
print(song1.duration)

print(song2.title)
print(song2.artist)
print(song2.duration)
```
