```python
food_apps = ["Zomato", "Swiggy", "Dominos", "Uber Eats", "EatSure"]

app_iterator = iter(food_apps)

print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
print(next(app_iterator))
```

```python
def playlist_generator(songs):
    for song in songs:
        yield song


songs = ["Believer", "Perfect", "Shape of You", "Faded"]

playlist = playlist_generator(songs)

print(next(playlist))
print(next(playlist))
print(next(playlist))
print(next(playlist))
```

```python
cart = ["Pizza", "Burger", "Fries", "Coke"]

for index, item in enumerate(cart, start=1):
    print(f"Item {index}: {item}")
```

```python
teams = ["MI", "CSK", "RCB", "PNJB"]
points = [18, 16, 14, 12]

for team, point in zip(teams, points):
    print(f"Team: {team}, Points: {point}")
```

```python
def order_id_generator():
    order_id = 1001

    while True:
        yield order_id
        order_id += 1


orders = order_id_generator()

print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
print(next(orders))
```
