# task 1

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


product1 = Product("T-Shirt", 1000)

print("Product:", product1.name)
print("Discounted Price:", product1.get_discounted_price())
```

# task 2

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


product1 = Product("T-Shirt", 1000)
product2 = Electronics("Headphones", 2000)

print(product1.get_discounted_price())
print(product2.get_discounted_price())
```

# task 3

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


def show_final_price(item):
    print("Product:", item.name)
    print("Final Price:", item.get_discounted_price())


product1 = Product("T-Shirt", 1000)
product2 = Electronics("Headphones", 2000)

show_final_price(product1)
show_final_price(product2)
```

# task 4

```python
class Ticket:
    def __init__(self, movie_name, price):
        self.movie_name = movie_name
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50


ticket1 = Ticket("Avengers", 200)
ticket2 = PremiumTicket("Avengers", 200)

print("Normal Ticket:", ticket1.get_final_price())
print("Premium Ticket:", ticket2.get_final_price())
```
