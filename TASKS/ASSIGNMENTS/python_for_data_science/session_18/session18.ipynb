# TASK 2

```python
import re
def has_valid_date(text):
    return bool(re.search(r'\d{2}/\d{2}/\d{4}', text))

print(has_valid_date("Meeting on 25/12/2025"))
```

# TASK 3

```python
import re
text = "Item A: Rs. 299, Item B: Rs. 1500, Item C: Rs. 850"
prices = re.findall(r'Rs\.\s(\d+)', text)
total = sum(int(p) for p in prices)
print(total) 
```
